```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error

# Load dataset
data_url = "https://cocl.us/concrete_data"
df = pd.read_csv(data_url)

# Split features and target
X = df.drop(columns=['Strength'])
y = df['Strength']

# Function to build a baseline model
def build_baseline_model():
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X.shape[1],)),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model

# Part A: Build a Baseline Model
mse_list = []
for _ in range(50):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=None)
    model = build_baseline_model()
    model.fit(X_train, y_train, epochs=50, verbose=0, validation_data=(X_test, y_test))
    y_pred = model.predict(X_test)
    mse_list.append(mean_squared_error(y_test, y_pred))

print("Baseline Model - Mean MSE:", np.mean(mse_list))
print("Baseline Model - Std Dev MSE:", np.std(mse_list))

# Part B: Normalize the Data
scaler = StandardScaler()
X_normalized = scaler.fit_transform(X)

mse_list = []
for _ in range(50):
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=None)
    model = build_baseline_model()
    model.fit(X_train, y_train, epochs=50, verbose=0, validation_data=(X_test, y_test))
    y_pred = model.predict(X_test)
    mse_list.append(mean_squared_error(y_test, y_pred))

print("Normalized Model - Mean MSE:", np.mean(mse_list))
print("Normalized Model - Std Dev MSE:", np.std(mse_list))

# Part C: Increase the Number of Epochs
mse_list = []
for _ in range(50):
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=None)
    model = build_baseline_model()
    model.fit(X_train, y_train, epochs=100, verbose=0, validation_data=(X_test, y_test))
    y_pred = model.predict(X_test)
    mse_list.append(mean_squared_error(y_test, y_pred))

print("100 Epochs Model - Mean MSE:", np.mean(mse_list))
print("100 Epochs Model - Std Dev MSE:", np.std(mse_list))

# Part D: Increase the Number of Hidden Layers
def build_deeper_model():
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X.shape[1],)),
        Dense(10, activation='relu'),
        Dense(10, activation='relu'),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model

mse_list = []
for _ in range(50):
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=None)
    model = build_deeper_model()
    model.fit(X_train, y_train, epochs=50, verbose=0, validation_data=(X_test, y_test))
    y_pred = model.predict(X_test)
    mse_list.append(mean_squared_error(y_test, y_pred))

print("Deeper Model - Mean MSE:", np.mean(mse_list))
print("Deeper Model - Std Dev MSE:", np.std(mse_list))
```
